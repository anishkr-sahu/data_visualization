```python
# data cleaning
data.isnull().sum()
```

```python
data['SepalLengthCm'] = data['SepalWidthCm'].fillna(data['SepalLengthCm'].mean())
data['SepalWidthCm'] = data['SepalWidthCm'].fillna(data['SepalWidthCm'].mean())
data['PetalLengthCm'] = data['PetalLengthCm'].fillna(data['PetalLengthCm'].mean())
data['PetalWidthCm'] = data['PetalWidthCm'].fillna(data['PetalWidthCm'].mean())
```

```python
# data visualization
import matplotlib.pyplot as plt
```

```python
# line plot : It tells the trends of the data
plt.plot(data['SepalLengthCm'])
plt.title("line plot")
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

```python
plt.plot(data['SepalWidthCm'])
plt.title('Line plot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

```python
plt.plot(data['Species'])
plt.title('Line plot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

```python
plt.scatter(data['SepalLengthCm'],data['SepalWidthCm'])
plt.title('scatter plot')
plt.xlabel('sepalLengthcm')
plt.ylabel('sepalwidthcm')
plt.show()
```

```python
plt.scatter(data['PetalLengthCm'],data['PetalWidthCm'])
plt.title('scatter plot')
plt.xlabel('petalLengthcm')
plt.ylabel('petalWidthcm')
plt.show()
```

```python
plt.scatter(data['PetalLengthCm'], data['Species'])
plt.title('scatter plot')
plt.xlabel('petalLengthcm')
plt.ylabel('species')
plt.show()
```

```python
plt.hist(data['PetalLengthCm'],bins=20)
plt.title('histogram')
plt.xlabel('petalLengthcm')
plt.ylabel('count')
plt.show()
```

```python
plt.hist(data['Species'],bins=20)
plt.title('histogram')
plt.xlabel('species')
plt.ylabel('count')
plt.show()
```

```python
data['Species'].value_counts().plot(kind='pie',autopct='%0.2f%%')
plt.title('pie chart')
plt.show()
```

```python
data['Species'].value_counts().plot(kind='bar')
plt.title('bar chart')
plt.xlabel('species')
plt.ylabel('count')
plt.show()
```

```python
plt.boxplot(data['SepalLengthCm'])
plt.title('boxplot')
plt.ylabel('sepalLengthcm')
plt.show()
```

```python
data.iloc[:,1:5].plot(kind='box')
plt.title("box plot")
plt.show()
```
